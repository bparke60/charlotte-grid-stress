# Gradient Descent from Scratch
Independent portfolio implementation motivated by gradient descent and LMS coursework.

```python
import numpy as np
from gradient_descent import fit_gradient_descent, predict
X = np.arange(20, dtype=float).reshape(-1, 1)
y = 3 + 2 * X[:, 0]
model = fit_gradient_descent(X, y)
print('First loss:', model['losses'][0])
print('Final loss:', model['losses'][-1])
```

```python
import matplotlib.pyplot as plt
plt.plot(model['losses'])
plt.xlabel('Epoch')
plt.ylabel('Training MSE')
plt.show()
```
